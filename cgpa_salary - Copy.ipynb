## First import library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import linear_model
```

## import data set

```python
df=pd.read_csv('cgpa_salary_data.csv')
```

```python
df=pd.DataFrame(df)
```

```python
df['Salary in $'] = df['Salary in $'].astype(int)
```

```python
df
```

## Data in Scatter Form



```python

plt.scatter(df['CGPA'],df['Salary in $'])
plt.xlabel('CGPA',color='green')
plt.ylabel('Salary in $',color='green')
plt.title('CGPA and SALARY($)',family='arial',color='red')

```

* This is a pandas function used for integer-location based indexing, meaning you can access rows and columns by their numerical index.

```python
X=df.iloc[:,0:1]
y=df.iloc[:,-1]
```

```python
X
```

```python
y
```

# There we Split data set

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
plt.scatter(df['CGPA'],df['Salary in $'])
plt.plot(X_train,lr.predict(X_train),color='red')
```

```python
from sklearn.metrics import  mean_absolute_error,mean_squared_error,r2_score
```

```python
y_pred=lr.predict(X_test)
```

```python
y_test.values
```

```python
print("MAE",mean_absolute_error(y_test,y_pred))
```

```python
lr.predict([[2.3]])
```

```python
print("MSE",mean_squared_error(y_test,y_pred))
```
